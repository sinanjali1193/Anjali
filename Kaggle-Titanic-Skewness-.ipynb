```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew

# Load dataset
df = sns.load_dataset("titanic")

# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64'])

# Statistics
print("===== STATISTICS =====")

for col in numeric_cols.columns:
    data = numeric_cols[col].dropna()

    print(f"\nColumn: {col}")
    print("Mean:", data.mean())
    print("Median:", data.median())
    print("Standard Deviation:", data.std())
    print("Variance:", data.var())
    print("Skewness:", skew(data))

# Histograms
numeric_cols.hist(figsize=(15,10), bins=20)
plt.tight_layout()
plt.show()
```

```python
for col in numeric_cols.columns:
    data = numeric_cols[col].dropna()
    skewness = skew(data)

    if skewness > 1:
        print(f"{col} is highly right skewed")
    elif skewness < -1:
        print(f"{col} is highly left skewed")
    else:
        print(f"{col} is approximately symmetric")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("titanic")

# Convert categorical sex column
df['sex'] = df['sex'].map({'male': 0, 'female': 1})

# Select numeric columns
corr_data = df[['survived', 'pclass', 'sex', 'age', 'fare', 'sibsp', 'parch']]

# Correlation matrix
corr_matrix = corr_data.corr()

print(corr_matrix)

# Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
